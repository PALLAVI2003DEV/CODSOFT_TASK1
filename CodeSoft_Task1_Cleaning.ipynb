```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/CodSoft_Task1/zomato.csv')

print("Shape:", df.shape)
print(df.info())
print(df.head())
print(df.columns.tolist())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
df.info()
display(df.head())
```

```python
print("=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Duplicates ===")
print("Total duplicates:", df.duplicated().sum())

print("\n=== Inconsistent Entries ===")
print(df['rate'].unique())
print(df['approx_cost(for two people)'].unique()[:20])
```

```python
df = df.drop(['url', 'address', 'phone', 'dish_liked', 'reviews_list', 'menu_item'], axis=1)
df = df.drop_duplicates()

def clean_rate(x):
    if pd.isna(x) or x in ['NEW', '-']:
        return np.nan
    return float(str(x).split('/')[0])

df['rate'] = df['rate'].apply(clean_rate)
```

```python
cols_to_drop = ['url', 'address', 'phone', 'dish_liked', 'reviews_list', 'menu_item']
df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])
df = df.drop_duplicates()

def clean_rate(x):
    if pd.isna(x) or x in ['NEW', '-']:
        return np.nan
    return float(str(x).split('/')[0])

df['rate'] = df['rate'].apply(clean_rate)
```

```python
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].astype(str).str.replace(',', '')
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')

df = df.rename(columns={
    'approx_cost(for two people)': 'cost_for_two',
    'listed_in(type)': 'type',
    'listed_in(city)': 'city'
})
```

```python
df['rate'] = df['rate'].fillna(df['rate'].median())
df['cost_for_two'] = df['cost_for_two'].fillna(df['cost_for_two'].median())
df['rest_type'] = df['rest_type'].fillna('Unknown')
df['cuisines'] = df['cuisines'].fillna('Unknown')
df = df.dropna(subset=['location'])
df['votes'] = df['votes'].astype(int)
```

```python
print("Final shape:", df.shape)
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
print(df.dtypes)
```

```python
df.to_csv('/content/drive/MyDrive/CodSoft_Task1/zomato_cleaned.csv', index=False)
print("Saved successfully! Task 1 done.")
```
